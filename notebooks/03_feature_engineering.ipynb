```python
service_columns = [
    "PhoneService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

data["Service_Count"] = 0

for col in service_columns:
    data["Service_Count"] += (
        data[col] == "Yes"
    ).astype(int)

    data["Estimated_Billed_Value"] = (
    data["MonthlyCharges"] * data["tenure"]
)

    data["Charge_per_Tenure_Month"] = (
    data["TotalCharges"]
    / data["tenure"].replace(0, np.nan)
)

    data["Tenure_Group"] = pd.cut(
    data["tenure"],
    bins=[-1, 6, 12, 24, 48, 72],
    labels=[
        "0-6",
        "7-12",
        "13-24",
        "25-48",
        "49-72"
    ]
)

    data["Streaming_Adoption"] = (
    (
        (data["StreamingTV"] == "Yes")
        |
        (data["StreamingMovies"] == "Yes")
    )
).astype(int)

    data.to_csv(
    "../data/processed/customer_master.csv",
    index=False
)
```
