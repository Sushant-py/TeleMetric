```python
import pandas as pd
import numpy as np

data = pd.read_csv(
    "../data/processed/customer_master.csv"
)

print(data.shape)
print(data.columns.tolist())

segment_data = data[
    [
        "customerID",
        "tenure",
        "MonthlyCharges",
        "Estimated_Billed_Value",
        "Service_Count",
        "Churn_Target"
    ]
].copy()

value_cutoff = segment_data[
    "Estimated_Billed_Value"
].median()

engagement_cutoff = segment_data[
    "tenure"
].median()

def assign_segment(row):
    
    high_value = (
        row["Estimated_Billed_Value"]
        >= value_cutoff
    )
    
    high_tenure = (
        row["tenure"]
        >= engagement_cutoff
    )

    if high_value and high_tenure:
        return "High Value - Established"
    
    elif high_value and not high_tenure:
        return "High Value - New"
    
    elif not high_value and high_tenure:
        return "Low Value - Established"
    
    else:
        return "Low Value - New"

segment_data["Customer_Segment"] = (
    segment_data.apply(assign_segment, axis=1)
)

segment_data.groupby(
    "Customer_Segment"
)["Churn_Target"].agg(
    ["count", "mean"]
)
```
